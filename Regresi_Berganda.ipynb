##1. Inisialisasi dan Pembuatan Dataset

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
from sklearn.linear_model import LinearRegression

# Membuat data frame dari data yang diberikan
data = {
    'Partisipan': np.arange(1, 11),
    'PMakan': [52, 48, 62, 58, 42, 68, 72, 78, 85, 88],
    'PTranspor': [18, 22, 28, 32, 45, 48, 52, 62, 68, 75],
    'BB': [58, 53, 67, 68, 60, 72, 82, 83, 88, 92]
}

df = pd.DataFrame(data)
df
```

##2. Pemodelan Regresi Linear Berganda

```python
# Menentukan variabel independen (X) dan dependen (y)
X = df[['PTranspor', 'PMakan']]
y = df['BB']

# Membuat dan melatih model regresi linear
model = LinearRegression()
model.fit(X, y)

# Mengambil koefisien regresi dan nilai R^2
intercept = model.intercept_
b1, b2 = model.coef_
r2 = model.score(X, y)

print(f"Intercept: {intercept:.3f}")
print(f"Koefisien PTranspor (b1): {b1:.3f}")
print(f"Koefisien PMakan (b2): {b2:.3f}")
print(f"R-squared (R2): {r2:.3f}")
```

##3. Pembentukan Meshgrid Surface 3D

```python
# Membuat grid untuk surface plane 3D
x1_range = np.linspace(df['PTranspor'].min(), df['PTranspor'].max(), 20)
x2_range = np.linspace(df['PMakan'].min(), df['PMakan'].max(), 20)
X1_grid, X2_grid = np.meshgrid(x1_range, x2_range)

# Menghitung bidang hasil persamaan regresi
Y_grid = intercept + (b1 * X1_grid) + (b2 * X2_grid)
```

##4. Visualisasi Grafik Regresi Linear Berganda 3D

```python
# Membuat Plot 3D Menggunakan Plotly
fig = go.Figure()

# 1. Menambahkan Scatter Plot Data Aktual (Titik Merah)
fig.add_trace(go.Scatter3d(
    x=df['PTranspor'],
    y=df['PMakan'],
    z=df['BB'],
    mode='markers',
    marker=dict(size=6, color='red', opacity=0.9),
    name='Data Aktual'
))

# 2. Menambahkan Surface Regresi Plane 3D
fig.add_trace(go.Surface(
    x=X1_grid,
    y=X2_grid,
    z=Y_grid,
    colorscale='Viridis',
    opacity=0.7,
    colorbar=dict(title='BB'),
    showscale=True
))

# 3. Format Layout dan Judul Sesuai Gambar
title_text = f"<b>REGRESI LINEAR BERGANDA 3D</b><br><sup>Ŷ = {intercept:.3f} + ({b1:.3f}) X₁ + ({b2:.3f}) X₂  |  R² = {r2:.3f}</sup>"

fig.update_layout(
    title=dict(
        text=title_text,
        x=0.5,
        xanchor='center',
        font=dict(size=14)
    ),
    scene=dict(
        xaxis_title='Pengeluaran Transportasi (X₁)',
        yaxis_title='Pengeluaran Makanan (X₂)',
        zaxis_title='BB(Y)',
        camera=dict(
            eye=dict(x=-1.5, y=-1.5, z=1.2)
        )
    ),
    width=800,
    height=600,
    margin=dict(l=0, r=0, b=0, t=60)
)

fig.show()
```
